# Contexto: física de la ecuación de Lane-Emden

La ecuación de Lane-Emden surge al estudiar la estructura interna de una estrella bajo el supuesto de equilibrio hidrostático. En este escenario, la configuración estelar está determinada por el balance entre dos efectos fundamentales: la atracción gravitatoria, que tiende a comprimir la materia hacia el centro, y el gradiente de presión, que actúa en sentido opuesto evitando el colapso.

## 1. Equilibrio hidrostático

El equilibrio mecánico de una estrella esférica se describe mediante:

$$\frac{dP}{dr} = -\frac{G M(r)}{r^2}\rho$$

Esta expresión indica que la presión disminuye con el radio para compensar la fuerza gravitatoria ejercida por la masa interna.

## 2. Conservación de la masa

La variación de la masa contenida dentro de una esfera de radio $r$ está dada por:

$$\frac{dM(r)}{dr} = 4\pi r^2 \rho$$

Esta ecuación relaciona directamente la distribución de densidad con la estructura global de la estrella.


## 3. Aproximación politrópica

Para cerrar el sistema de ecuaciones, se adopta una relación simplificada entre presión y densidad:

$$P = K \rho^{1 + \frac{1}{n}}$$

donde $K$ es una constante y $n$ es el índice politrópico. Esta aproximación permite modelar distintos regímenes físicos del material estelar con un único parámetro.

## 4. Variables adimensionales

Con el objetivo de eliminar constantes físicas y obtener una formulación universal, se introducen las variables adimensionales:

$$\rho = \rho_c \theta^n, \qquad r = \alpha \xi$$

donde $\rho_c$ es la densidad central y $\alpha$ es una escala característica del sistema. Este cambio simplifica considerablemente las ecuaciones del problema.

## 5. Ecuación de Lane-Emden

Tras sustituir las variables adimensionales en las ecuaciones de equilibrio y masa, se obtiene la ecuación diferencial de Lane-Emden:

$$\frac{1}{\xi^2} \frac{d}{d\xi} \left( \xi^2 \frac{d\theta}{d\xi} \right) = -\theta^n$$

Esta ecuación describe la estructura radial de una estrella politrópica de manera completamente adimensional.

## 6. Interpretación física

La función $\theta(\xi)$ representa la variación de la densidad en el interior de la estrella. El punto donde $\theta(\xi_1)=0$ define el borde estelar en unidades adimensionales, permitiendo determinar propiedades globales como el radio, la masa y la concentración de densidad.


En este proyecto se resolverá esta ecuación numéricamente utilizando el método de Runge-Kutta de cuarto orden (RK4).

# Proceso: Metodología numérica y enfoque del proyecto

A partir del marco teórico descrito anteriormente, el objetivo de este trabajo consistió en resolver numéricamente la ecuación de Lane-Emden para distintos valores del índice politrópico ($n$ ). Dado que esta ecuación diferencial de segundo orden no admite soluciones analíticas generales, fue necesario reformularla como un sistema de ecuaciones diferenciales de primer orden y abordarla mediante métodos de integración numérica.

En particular, se implementó el método de Runge-Kutta de cuarto orden (RK4), el cual permite obtener soluciones estables y precisas a lo largo del dominio radial adimensional ( $\xi$ ). Para evitar la singularidad en el origen (( $\xi = 0$ )), se utilizaron condiciones iniciales obtenidas a partir de una expansión en serie de Taylor, lo que garantiza un inicio consistente de la integración cerca del centro estelar.

Durante la integración, se determinó el punto ( $\xi_1$ ) en el cual la función ( $\theta(\xi)$ ) se anula, identificando así el radio adimensional de la estrella. Este valor, junto con el gradiente en la superficie, permitió calcular magnitudes físicas relevantes como la masa adimensional, la razón de concentración de densidad ( $\rho_c / \rho_m$ ) y la presión central.

Una vez obtenidas las soluciones para distintos valores de ( $n$ ), se analizaron los perfiles de densidad resultantes y se compararon con un modelo solar estándar. Esta comparación permitió evaluar el alcance y las limitaciones de la aproximación politrópica, especialmente en lo que respecta a la descripción del núcleo estelar, donde los efectos físicos reales se apartan de las simplificaciones asumidas en el modelo.

De esta manera, el proyecto conecta directamente la formulación teórica de la ecuación de Lane-Emden con su implementación computacional y su interpretación física, permitiendo no solo obtener soluciones numéricas, sino también analizar críticamente su validez en el contexto de la estructura estelar real.

# Discusión: discrepancias en la concentración de densidad 

Basandonos en la segunda gráfica generada por el código donde comparamos directamente el perfil de densidad polítropo $n = 3$ (modelo de Eddington) y el modelo solar estándar de Christensen‑Dalsgaard observamos unas diferencias significativas, particularmente en el interior profundo.

### En el núcleo: 

En la parte más interna $\frac{r}{R} < 0.2$ la densidad del modelo solar real es apreciablemente mayor que la predicha por el polítropo

### En la envoltura u exterior:

En la parte más externa $\frac{r}{R} > 0.2$ la situación se invierte: el polítropo sobrestima ligeramente la densidad respecto al Sol real. Esto es una consecuencia de la conservación de la masa total; si el centro es menos denso, las capas externas deben serlo más para alcanzar la misma masa media.

### Razones: 

El modelo politrópico supone una relación funcional de la forma P ∝ ρ^(1+1/n) con un índice n constante en toda la estrella. En el caso n = 3, esta relación corresponde a una esfera de gas ideal con transporte radiativo dominante (Eddington). Sin embargo, el interior solar real se aparta de esta simplificación por varias razones:

1. Composición química variable: En el núcleo solar la fracción de helio es mayor debido al quemado del hidrógeno durante los ∼4.6 Gyr de evolución. El peso molecular medio μ aumenta, y mediante la ecuación de estado del gas ideal (P ∝ ρT/μ), una misma presión se consigue con mayor densidad. Un polítropo simple no puede capturar este gradiente de composición. Así mismo en las en zonas más internas se tiene más helio (producto de la fusión) y en las zonas más exteriores se tiene más hidrógeno, estas diferencias en la composición modifican la relación presión-densidad.

2. Procesos de transporte de energía: Mientras que en la zona radiativa interior (0 < r < 0.7 R⊙) la aproximación n = 3 es plausible, en la zona convectiva externa la relación presión‑densidad cambia (el índice politrópico efectivo local es menor). Un modelo de índice único promedia comportamientos y no reproduce correctamente la concentración central.

3. Presión de radiación y degeneración parcial: En el núcleo solar las temperaturas son suficientemente altas para que la presión de radiación no sea despreciable, lo que modifica el exponente politrópico efectivo. Además, en estrellas ligeramente más masivas o en etapas posteriores, la degeneración electrónica puede influir.

4. Física no politrópica: La producción de energía nuclear aumenta temperatura local (y además es sensible a la temperatura ciclo pp) y crea un perfil de temperatura y densidad que no obedece estrictamente a P ∝ ρ^γ. El acoplamiento entre la ecuación de estado, la opacidad y la generación de energía requiere un tratamiento autoconsistente que el modelo politrópico no ofrece.

Estas razones no solo describen las discrepancias en el interior si no también justifican el modelo en las zonas exteriores; en las capas externas se tiene una menor temperatura, menor generación nuclear y un comportamiento del gas más “ideal”, es decir, todas las razones anteriormente mencionadas son menos dominantes y la aproximación encaja mucho mejor con el modelo solar.
