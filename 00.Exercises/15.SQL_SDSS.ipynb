# Preguntas de SQL

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("shows.db")
```

```python
#Pregunta 1: ¿Cuántas series tienen una calificación perfecta de 10.0?

query = """
SELECT COUNT(*)
FROM ratings
WHERE rating = 10.0;
"""

pd.read_sql_query(query, conn)
```

```python
#Pregunta 2: ¿Cuántos episodios tiene Black Mirror?

query = """
SELECT episodes
FROM shows
WHERE title = 'Black Mirror';
"""

pd.read_sql_query(query, conn)
```

```python
#Pregunta 3: ¿Cuántas series existen en el género ciencia ficción?

query = """
SELECT COUNT(DISTINCT show_id)
FROM genres
WHERE genre = 'Sci-Fi';
"""

pd.read_sql_query(query, conn)
```

```python
#Pregunta 4: ¿Cuál es la serie mejor valorada del género de terror?

query = """
SELECT s.title, r.rating
FROM shows s
JOIN genres g ON s.id = g.show_id
JOIN ratings r ON s.id = r.show_id
WHERE g.genre = 'Horror'
ORDER BY r.rating DESC
LIMIT 1;
"""

pd.read_sql_query(query, conn)
```

```python
#Pregunta 5: ¿Cuántas series de animación hay en la base de datos?

query = """
SELECT COUNT(DISTINCT show_id) AS n_animation
FROM genres
WHERE genre = 'Animation';
"""

pd.read_sql_query(query, conn)
```

```python
#Pregunta 6: ¿Cuáles fueron las 10 series de animación peor valoradas entre 2005 y 2010?

query = """
SELECT s.title,
       s.year,
       r.rating
FROM shows s
JOIN genres g ON s.id = g.show_id
JOIN ratings r ON s.id = r.show_id
WHERE g.genre = 'Animation'
  AND s.year BETWEEN 2005 AND 2010
ORDER BY r.rating ASC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

# Preguntas de SDSS

```python
from astroquery.sdss import SDSS
import pandas as pd
```

```python
#Pregunta 1: Encuentra galaxias cerca de RA=141 más brillantes que g=18 y recupera ObjID, RA, DEC y las cinco magnitudes.

query1 = """
SELECT TOP 50
    p.objID,
    p.ra,
    p.dec,
    p.u,
    p.g,
    p.r,
    p.i,
    p.z
FROM PhotoObj AS p
WHERE
    ABS(p.ra - 141.0) < 0.5
    AND p.g < 18
ORDER BY p.g ASC
"""

result1 = SDSS.query_sql(query1)

df1 = result1.to_pandas()
df1.head()
```

```python
# Pregunta 2: Busca cuásares con espectros donde z > 4.5 y zWarning=0. ¿Cuál tiene el mayor corrimiento al rojo?

query2 = """
SELECT TOP 1
    s.specObjID,
    s.ra,
    s.dec,
    s.z
FROM SpecObj AS s
WHERE
    s.class = 'QSO'
    AND s.z > 4.5
    AND s.zWarning = 0
ORDER BY s.z DESC
"""

result2 = SDSS.query_sql(query2)

df2 = result2.to_pandas()
df2
```

```python
#Pregunta 3: Devuelve solo galaxias entre RA=140.25–140.75 y DEC=20.25–20.75

query3 = """
SELECT TOP 100
    p.objID,
    p.ra,
    p.dec
FROM PhotoObj AS p
WHERE
    p.type = 3
    AND p.ra BETWEEN 140.25 AND 140.75
    AND p.dec BETWEEN 20.25 AND 20.75
"""

result3 = SDSS.query_sql(query3)

df3 = result3.to_pandas()
df3.head()
```
