## 2. Método de Bisección 

Sea:
$$f(x)=x^3 - 0.165x^2 + 3.993\times 10^{-4}$$

**Raíz 1: $x \in (-0.05,\,-0.02)$**

**Iteración 1:**

$$x_l=-0.05,\quad x_u=-0.02$$

$$x_m=\frac{x_l+x_u}{2}=\frac{-0.05-0.02}{2}=-0.035$$

Evaluaciones:

$$f(x_l)=f(-0.05)\approx -2.25\times10^{-4}$$

$$f(x_m)=f(-0.035)\approx -4.29\times10^{-5}$$

$$f(x_u)=f(-0.02)\approx 2.93\times10^{-4}$$

Productos:

$$f(x_l)f(x_m)>0$$

$$f(x_m)f(x_u)<0$$

Así nuestro nuevo intervalo es: $$[-0.035,\,-0.02]$$

**Iteración 2:**

$$x_l=-0.035,\quad x_u=-0.02$$

$$x_m=\frac{-0.035-0.02}{2}=-0.0275$$

$$f(x_l)\approx -4.29\times10^{-5}$$

$$f(x_m)=f(-0.0275)\approx 1.33\times10^{-4}$$

$$f(x_l)f(x_m)<0$$
$$f(x_m)f(x_u)>0$$

Así nuestro nuevo intervalo es: $$[-0.035,\,-0.0275]$$

## 2. Método de Newton-Raphson

Sea:

$$f(x)=x^3 - 0.165x^2 + 3.993\times 10^{-4}$$

Derivada:

$$f'(x)=3x^2 - 0.33x$$

Fórmula:

$$x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}$$

---

**Raíz 1 (semilla: $x_0=-0.035$)**

**Iteración 1:**

$$x_0=-0.035$$

$$f(x_0)\approx -4.29\times10^{-5}$$

$$f'(x_0)=3(-0.035)^2-0.33(-0.035)\approx 0.0152$$

$$x_1=x_0-\frac{f(x_0)}{f'(x_0)}
=-0.035-\frac{-4.29\times10^{-5}}{0.0152}
\approx -0.03218$$

Error relativo aproximado:

$$\varepsilon_a=\left|\frac{x_1-x_0}{x_1}\right|\times100
\approx 8.76\%$$

Dígitos significativos correctos; ninguno :(

**Iteración 2:**

$$x_1=-0.03218$$

$$f(x_1)\approx -2.62\times10^{-6}$$

$$f'(x_1)\approx 0.0136$$

$$x_2=x_1-\frac{f(x_1)}{f'(x_1)}
\approx -0.03199$$

$$\varepsilon_a\approx 0.59\%$$

$$1 \text{ dígito significativo correcto}$$

**Iteración 3:**

$$x_2=-0.03199$$

$$f(x_2)\approx -1.05\times10^{-8}$$

$$f'(x_2)\approx 0.0135$$

$$x_3\approx -0.03199$$

$$\varepsilon_a\approx 0.003\%$$

$$\text{Al menos 3 dígitos significativos correctos}$$
