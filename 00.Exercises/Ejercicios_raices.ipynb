## 2. Método de Newton-Raphson

Sea:

$$f(x)=x^3 - 0.165x^2 + 3.993\times 10^{-4}$$

Derivada:

$$f'(x)=3x^2 - 0.33x$$

Fórmula:

$$x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}$$

**Raíz 1 (semilla: $x_0=-0.035$)**

**Iteración 1:**

$$x_0=-0.035$$

$$f(x_0)\approx -4.29\times10^{-5}$$

$$f'(x_0)=3(-0.035)^2-0.33(-0.035)\approx 0.0152$$

$$x_1=x_0-\frac{f(x_0)}{f'(x_0)}
=-0.035-\frac{-4.29\times10^{-5}}{0.0152}
\approx -0.03218$$

Error relativo aproximado:

$$\varepsilon_a=\left|\frac{x_1-x_0}{x_1}\right|\times100
\approx 8.76\%$$

Dígitos significativos correctos; ninguno :(

**Iteración 2:**

$$x_1=-0.03218$$

$$f(x_1)\approx -2.62\times10^{-6}$$

$$f'(x_1)=3(-0.03218)^2-0.33(-0.03218)\approx 0.0136$$

$$x_2=x_1-\frac{f(x_1)}{f'(x_1)}
=-0.03218-\frac{-2.62\times10^{-6}}{0.0136}
\approx -0.03199$$

Error relativo aproximado:

$$\varepsilon_a=\left|\frac{x_2-x_1}{x_2}\right|\times100
\approx 0.59\%$$

$$1 \text{ dígito significativo correcto}$$

**Iteración 3:**

$$x_2=-0.03199$$

$$f(x_2)\approx -1.05\times10^{-8}$$

$$f'(x_2)=3(-0.03199)^2-0.33(-0.03199)\approx 0.0135$$

$$x_3=x_2-\frac{f(x_2)}{f'(x_2)}
=-0.03199-\frac{-1.05\times10^{-8}}{0.0135}
\approx -0.03199$$

Error relativo aproximado:

$$\varepsilon_a=\left|\frac{x_3-x_2}{x_3}\right|\times100
\approx 0.003\%$$

## 3. Método Secante

Sea:
$$f(x)=x^3 - 0.165x^2 + 3.993\times 10^{-4}$$

**Raíz 1: $x \in (-0.05,\,-0.02)$**

**Iteración 1:**

$$x_0=-0.05,\quad x_1=-0.02$$

$$f(x_0)\approx -2.25\times10^{-4}, \quad f(x_1)\approx 2.93\times10^{-4}$$

$$x_2=x_1-\frac{f(x_1)(x_1-x_0)}{f(x_1)-f(x_0)}$$

$$x_2\approx -0.02 - \frac{(2.93\times10^{-4})(-0.02+0.05)}{(2.93\times10^{-4})-(-2.25\times10^{-4})}$$

$$x_2\approx -0.03696$$


**Iteración 2:**

$$x_1=-0.02,\quad x_2=-0.03696$$

$$f(x_2)\approx -2.02\times10^{-5}$$

$$x_3=x_2-\frac{f(x_2)(x_2-x_1)}{f(x_2)-f(x_1)}$$

$$x_3\approx -0.03696 - \frac{(-2.02\times10^{-5})(-0.03696+0.02)}{(-2.02\times10^{-5})-(2.93\times10^{-4})}$$

$$x_3\approx -0.03574$$

Error relativo aproximado:

$$\varepsilon_a=\left|\frac{x_3-x_2}{x_3}\right|\times100 \approx 3.41\%$$

Dígitos significativos correctos: 1


**Iteración 3:**

$$x_2=-0.03696,\quad x_3=-0.03574$$

$$f(x_3)\approx -1.68\times10^{-6}$$

$$x_4=x_3-\frac{f(x_3)(x_3-x_2)}{f(x_3)-f(x_2)}$$

$$x_4\approx -0.03574 - \frac{(-1.68\times10^{-6})(-0.03574+0.03696)}{(-1.68\times10^{-6})-(-2.02\times10^{-5})}$$

$$x_4\approx -0.03564$$

Error relativo aproximado:

$$\varepsilon_a\approx 0.28\%$$

## Segundo Ejercicio: 

Para este ejercicio se nos piden solucionar el conjunto de ecuaciones: 

$$
\begin{cases}
f_1(x,y)=xy-2=0 \\
f_2(x,y)=x^2+y-5=0
\end{cases}
$$

Empleando el método de Newton-Raphson con los valores iniciales $(x_i , y_i)=(3,4)$ e iterar hasta que el error relativo sea menor a 0.05.

Vector de funciones:

$$
\mathbf{F}(x,y)=
\begin{pmatrix}
xy-2 \\
x^2+y-5
\end{pmatrix}
$$

Jacobiano:

$$
J(x,y)=
\begin{pmatrix}
\frac{\partial f_1}{\partial x} & \frac{\partial f_1}{\partial y} \\
\frac{\partial f_2}{\partial x} & \frac{\partial f_2}{\partial y}
\end{pmatrix}
=
\begin{pmatrix}
y & x \\
2x & 1
\end{pmatrix}
$$

Fórmula:

$$
\begin{pmatrix}
x_{n+1} \\
y_{n+1}
\end{pmatrix}
=
\begin{pmatrix}
x_n \\
y_n
\end{pmatrix}
-
J^{-1}(x_n,y_n)\mathbf{F}(x_n,y_n)
$$

**Iteración 1:**

$$x_0=3,\quad y_0=4$$

Evaluamos:

$$f_1(3,4)=3\cdot4-2=10$$

$$f_2(3,4)=3^2+4-5=8$$

Jacobiano:

$$
J(3,4)=
\begin{pmatrix}
4 & 3 \\
6 & 1
\end{pmatrix}
$$

Determinante:

$$\det J=4(1)-3(6)=-14$$

Inversa:

$$
J^{-1}=\frac{1}{-14}
\begin{pmatrix}
1 & -3 \\
-6 & 4
\end{pmatrix}
$$

Corrección:

$$
J^{-1}\mathbf{F}=
\frac{1}{-14}
\begin{pmatrix}
1 & -3 \\
-6 & 4
\end{pmatrix}
\begin{pmatrix}
10 \\
8
\end{pmatrix}
=
\frac{1}{-14}
\begin{pmatrix}
-14 \\
-28
\end{pmatrix}
=
\begin{pmatrix}
1 \\
2
\end{pmatrix}
$$

Actualización:

$$
\begin{pmatrix}
x_1 \\
y_1
\end{pmatrix}
=
\begin{pmatrix}
3 \\
4
\end{pmatrix}
-
\begin{pmatrix}
1 \\
2
\end{pmatrix}
=
\begin{pmatrix}
2 \\
2
\end{pmatrix}
$$

Error relativo:

$$
\varepsilon_a^{(x)}=\left|\frac{2-3}{2}\right|\times100=50\%
\quad,\quad
\varepsilon_a^{(y)}=\left|\frac{2-4}{2}\right|\times100=100\%
$$

$$\text{0 dígitos significativos correctos}$$
