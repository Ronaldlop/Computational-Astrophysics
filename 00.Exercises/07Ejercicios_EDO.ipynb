# 1. Ecuacicón Diferencial Ordinaria: 

Para resolver la siguiente EDO con las siguientes condiciones:

$$2\frac{\text{d}^2y}{\text{d}x^2} + 6\frac{\text{d}y}{\text{d}x} + \frac{25}{8}y = \cos x$$ $$y(0) = 5$$  $$y'(x=0) = 3$$

debemos hallar la solución homogénea, posteriormente la solución particular y por último aplicar las condiciones dadas. 

#### 1.1 Solución Homogénea: 

$$y''+3y'+\frac{25}{16}y=0$$
Proponemos: 
$$y=e^{rx}$$
Sacamos las derivadas respectivas: 
$$y'=re^{rx},\qquad y''=r^2e^{rx}$$
Sacamos factor común y obtenemos la ecuación carácteristica: 
$$r^2e^{rx}+3re^{rx}+\frac{25}{16}e^{rx}=0$$
$$r^2+3r+\frac{25}{16}=0$$
Ahora, con la ecuación carácteristica resolvemos la cuadrática asociada: 
$$r=\frac{-3\pm\sqrt{3^2-4\left(1\right)\left(\frac{25}{16}\right)}}{2}$$
$$r=\frac{-3\pm\sqrt{9-\frac{25}{4}}}{2}$$
$$r=\frac{-3\pm\sqrt{\frac{11}{4}}}{2}$$
$$r=\frac{-3\pm \frac{\sqrt{11}}{2}}{2}$$
$$r=-\frac{3}{2}\pm \frac{\sqrt{11}}{4}$$
Aplicamos la raices y hallamos la solución homogénea: 
$$r_1=-\frac{3}{2}+\frac{\sqrt{11}}{4}$$
$$r_2=-\frac{3}{2}-\frac{\sqrt{11}}{4}$$
$$y_h=C_1e^{\left(-\frac{3}{2}+\frac{\sqrt{11}}{4}\right)x}+C_2e^{\left(-\frac{3}{2}-\frac{\sqrt{11}}{4}\right)x}$$

### 1.2 Solución Particular: 

$$\frac{1}{2}\cos x$$
proponemos entonces: 
$$y_p=A\cos x+B\sin x$$
Sacamos las derivadas respectivas: 
$$y_p'=-A\sin x+B\cos x$$
$$y_p''=-A\cos x-B\sin x$$
Sustituimos en la ecuación y hallamos los coeficientes: 
$$y''+3y'+\frac{25}{16}y=\frac{1}{2}\cos x$$
$$(-A\cos x-B\sin x)+3(-A\sin x+B\cos x)+\frac{25}{16}(A\cos x+B\sin x)=\frac{1}{2}\cos x$$
Para los coeficiente de \(\cos x\)
$$-A+3B+\frac{25}{16}A$$
$$=\left(-1+\frac{25}{16}\right)A+3B$$
$$=\frac{9}{16}A+3B$$
Para los coeficiente de \(\sin x\)
$$-B-3A+\frac{25}{16}B$$
$$=\left(-1+\frac{25}{16}\right)B-3A$$
$$=\frac{9}{16}B-3A$$
Entonces obtenemos el sistema:
$$\frac{9}{16}A+3B=\frac{1}{2}$$
$$\frac{9}{16}B-3A=0$$
De la segunda:
$$\frac{9}{16}B=3A$$
$$9B=48A$$
$$B=\frac{16}{3}A$$
Sustituimos en la primera:
$$\frac{9}{16}A+3\left(\frac{16}{3}A\right)=\frac{1}{2}$$
$$\frac{9}{16}A+16A=\frac{1}{2}$$
$$\frac{9A+256A}{16}=\frac{1}{2}$$
$$\frac{265A}{16}=\frac{1}{2}$$
$$A=\frac{8}{265}$$
Ahora:
$$B=\frac{16}{3}\cdot \frac{8}{265}$$
$$B=\frac{128}{795}$$
Hallamos la solución particular: 
$$y_p=\frac{8}{265}\cos x+\frac{128}{795}\sin x$$


### 1.3 Solución General y Condiciones: 

la solución general es como ya mencionamos anteriormente la suma de la solución homogénea y la solución particular: 
$$y=C_1e^{\left(-\frac{3}{2}+\frac{\sqrt{11}}{4}\right)x}
+C_2e^{\left(-\frac{3}{2}-\frac{\sqrt{11}}{4}\right)x}
+\frac{8}{265}\cos x+\frac{128}{795}\sin x$$
Ahora a esta le aplicamos las condiciones iniciales: 
Primera condición: $(y(0)=5)$
Sustituimos $(x=0)$:
$$e^0=1,\qquad \cos 0=1,\qquad \sin 0=0$$
$$C_1+C_2+\frac{8}{265}=5$$
$$C_1+C_2=\frac{1317}{265}$$
Segunda condición:$(y'(0)=3$)
Derivamos la solución general:
$$y'= r_1C_1e^{r_1x}+r_2C_2e^{r_2x} -\frac{8}{265}\sin x+\frac{128}{795}\cos x$$
En (x=0):
$$r_1C_1+r_2C_2+\frac{128}{795}=3$$
$$r_1C_1+r_2C_2=\frac{2257}{795}$$
Resolviendo el sistema con
$$C_1+C_2=\frac{1317}{265}$$
se obtiene:
$$C_1=\frac{1317}{530}+\frac{16367\sqrt{11}}{8745}$$
$$C_2=\frac{1317}{530}-\frac{16367\sqrt{11}}{8745}$$

### 1.4 Solución final:  
$$y(x)=
C_1e^{\left(-\frac{3}{2}+\frac{\sqrt{11}}{4}\right)x}
+C_2e^{\left(-\frac{3}{2}-\frac{\sqrt{11}}{4}\right)x}
+\frac{8}{265}\cos x +\frac{128}{795}\sin x$$

$$C_1=\frac{1317}{530}+\frac{16367\sqrt{11}}{8745}$$
$$C_2=\frac{1317}{530}-\frac{16367\sqrt{11}}{8745}$$

# 2. Ecuacicón Diferencial de Preferencia (la ley de enfriamiento de Newton): 

Una taza de café está a $(80^\circ C)$ y se deja en una habitación a $(20^\circ C)$
La ley de enfriamiento de Newton dice:
$$\frac{dT}{dt}=-k(T-T_a)$$
donde:
- $T(t)$: temperatura del objeto
- $T_a$: temperatura ambiente
- k: constante de enfriamiento
Tomemos:
$$\frac{dT}{dt}=-0.2(T-20)$$
con
$$T(0)=80$$
Queremos aproximar la temperatura a los 3 minutos usando Euler. Para ello recordemos la formula: 
$$T_{n+1}=T_n+h f(t_n,T_n)$$
donde:
$$f(t,T)=-0.2(T-20)$$
Tomemos paso:
$$h=1$$


**Instante cero:** 
$$t_0=0,\qquad T_0=80$$
Evaluamos:
$$f(t_0,T_0)=-0.2(80-20)$$
$$f(t_0,T_0)=-12$$
Entonces:
$$T_1=T_0+h f(t_0,T_0)$$
$$T_1=80+1(-12)$$
$$T_1=68$$

**Instante uno:** 
$$t_1=1,\qquad T_1=68$$
$$f(t_1,T_1)=-0.2(68-20)$$
$$f(t_1,T_1)=-9.6$$
$$T_2=68+1(-9.6)$$
$$T_2=58.4$$

**Instante dos:** 
$$t_2=2,\qquad T_2=58.4$$
$$f(t_2,T_2)=-0.2(58.4-20)$$
$$f(t_2,T_2)=-7.68$$
$$T_3=58.4-7.68$$
$$T_3=50.72$$

**Resultado final:**
$$T(3)\approx 50.72^\circ C$$
